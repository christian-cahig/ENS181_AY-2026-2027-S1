```python
import math as mt
```

## Scenario 1

1. $1$, $1$, $1$, $1 \pm j3$
2. $0.5$, $-0.5$, $\pm j0.5$
3. $1$, $1$, $1$
4. $0$, $\pm j$

## Scenario 2

1. $x \left(y\right) = e^{y} \left(1 + \cos 3y \right)$
2. $z \left(x\right) = e^{0.5x} + \sin 0.5x$
3. $y \left(s\right) = e^{s} \left(1 + s^{2}\right)$
4. $w \left(t\right) = 1 + \cos t$

```python
def x(y): return mt.exp(y) * (1 + mt.cos(3*y))
def z(x): return mt.exp(x/2) + mt.sin(x/2)
def y(s): return mt.exp(s) * (1 + s**2)
def w(t): return 1 + mt.cos(t)
```

```python
indyvars = [-1.70, 1.19, -2.14, 0.67]
funcs = [x, z, y, w]

for i, (indy, func) in enumerate(zip(indyvars, funcs)):
    print(f"Item {i+1}. {func(indy)}")
```

## Scenario 3

1. $x \left(y\right) = 1.7 \left(e^{y} - 1\right)$
2. $z \left(x\right) = \frac{1}{8} x e^{0.5x}$
3. $y \left(s\right) = \left(13 + \frac{1}{6} s^{3}\right) e^{s} + s - 13$
4. $w \left(t\right) = - \sin t \ln \left(\sec t + \tan t\right) - \ln \left(\cos t\right)$

```python
def x(y): return 1.7 * (mt.exp(y) - 1)
def z(x): return x * mt.exp(x/2) / 8
def y(s): return s - 13 + ((13 + (s**3 / 6)) * mt.exp(s))
def w(t): return -mt.log(mt.cos(t)) - (mt.sin(t) * mt.log((1 + mt.sin(t)) / mt.cos(t)))
```

```python
indyvars = [1.70, -1.19, 2.14, -0.67]
funcs = [x, z, y, w]

for i, (indy, func) in enumerate(zip(indyvars, funcs)):
    print(f"Item {i+1}. {func(indy)}")
```
