```python
import math as mt

import numpy as np
```

## Scenario 1-A

```python
def y2(x): return (5*x) - (x**2)
```

```python
for i, x in enumerate([x for x in np.arange(0.25, stop=2.25, step=0.25) if not (x == 1.0)]):
    print(f"Question {i+1}: {y2(x)}")
```

## Scenario 1-B

```python
def y2(x): return (1.5 / (x**2)) - ((x**2) / 2)
```

```python
for i, x in enumerate([x for x in np.arange(0.25, stop=2.25, step=0.25) if not (x == 1.0)]):
    print(f"Question {i+1}: {y2(x)}")
```

## Scenario 2-A

```python
def y(x): return x / (mt.atan(x) + (4 / mt.pi) - (mt.pi / 4))
```

```python
for i in range(3):
    print(f"Question {i+1}: {y((i+1) * mt.pi / 6)}")
```

## Scenario 2-B

```python
def y(x): return mt.sqrt((mt.atan(x) + (mt.pi**2 / 16) - (mt.pi / 4)) / x)
```

```python
for i in range(3):
    print(f"Question {i+1}: {y((i+1) * mt.pi / 6)}")
```

## Scenario 3-A

```python
def T(t): return 85 - (60 * mt.exp(t * mt.log(7/12) / 2))
```

```python
for i, t in enumerate(np.arange(1.5, stop=6.5, step=1.0)):
    print(f"Question {i+1}. {T(t)}")
```

## Scenario 3-B

```python
for i, t in enumerate(np.arange(1.5, stop=6.5, step=1.0)):
    print(f"Question {i+1}. {T(t)}")
```

## Scenario 4-A

```python
def A(t): return 1600 / mt.cbrt(512 + (3*t))
t_full = (1000 - 512) / 3
```

```python
for i, t in enumerate([0.25, 0.50, 1.00]):
    print(f"Question {i+1}. {A(60 * t)}")
print(f"Question {4}. {t_full}")
print(f"Question {5}. {A(t_full)}")
```

## Scenario 4-B

```python
def A(t): return 0.2 * mt.pow(100 - t, 1.5)
t_empt = 400 / 4
```

```python
for i, t in enumerate([0.25, 0.50, 1.00]):
    print(f"Question {i+1}. {A(60 * t)}")
print(f"Question {4}. {t_empt}")
print(f"Question {5}. {A(t_empt - 10)}")
```

## Scenario 5-A

```python
def v(t):
    c = np.array([11, -16]); r = np.array([-2, -4])
    return (c * np.exp(r * t)).sum().item()
```

```python
for i, t in enumerate([0.1] + [k.item() for k in np.arange(0.25, stop=1.25, step=0.25)]):
    print(f"Question {i+1}. {v(t)}")
```

## Scenario 5-B

```python
def v(t): return ((500 * t) - 10) * mt.exp(-2*t)
```

```python
for i, t in enumerate([0.1] + [k.item() for k in np.arange(0.25, stop=1.25, step=0.25)]):
    print(f"Question {i+1}. {v(t)}")
```
