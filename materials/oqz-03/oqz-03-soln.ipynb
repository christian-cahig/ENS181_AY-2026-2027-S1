```python
import math as mt

import numpy as np
```

## Part I

```python
C_c = 1.5
C_s = 8.0
C_h = 450

R_cs = 0.1
R_sh = 0.2
R_ha = 0.4

Q_c = 100
Q_s = 0
Q_h = 0
```

```python
def T_a(t):
    return 25.0
```

```python
lam_0 =  0.0
lam_1 = -8.0224365
lam_2 = -0.5304573
lam_3 = -0.0054395
lams = np.array([lam_0, lam_1, lam_2, lam_3])
```

### Scenario 1

```python
def T_s(t):
    coeffs = np.array([85, 1.384, -20.520, -40.864])
    return (coeffs * np.exp(lams * t)).sum()
```

```python
def p(t):
    return 1 / (R_cs * C_c)

def q(t):
    return (Q_c / C_c) + (T_s(t) / (R_cs * C_c))
```

```python
print(f"Question 1. {p(0):.7f}")
for i, lam in enumerate(lams[1:], start=2):
    print(f"Question {i}. {p(-1 / lam):.7f}")

print(f"Question 5. {p(17):.7f}")

print(f"Question 6. {q(0):.7f}")
for i, lam in enumerate(lams[1:], start=7):
    print(f"Question {i}. {q(-1 / lam):.7f}")

print(f"Question 10. {q(17):.7f}")

print(f"Question 11. {T_a(0):.7f}")
```

### Scenario 2

```python
def r(t):
    return (1 / (R_sh * C_h)) + (1 / (R_ha * C_h))

def v(t):
    return (Q_h / C_h) + (T_s(t) / (R_sh * C_h)) + (T_a(t) / (R_ha * C_h))
```

```python
print(f"Question 1. {r(0):.7f}")
for i, lam in enumerate(lams[1:], start=2):
    print(f"Question {i}. {r(-1 / lam):.7f}")

print(f"Question 5. {r(17):.7f}")

print(f"Question 6. {v(0):.7f}")
for i, lam in enumerate(lams[1:], start=7):
    print(f"Question {i}. {v(-1 / lam):.7f}")

print(f"Question 10. {v(17):.7f}")

print(f"Question 11. {T_a(0):.7f}")
```

## Part II

### Scenario 3

```python
tau = 0.002
```

```python
def v_E(t):
    return 1 + mt.exp(-t / 0.002)

def i_E(t):
    return -mt.exp(-t / 0.002) / 2

def i_s(t):
    return 1 + i_E(t)
```

```python
for i, m in enumerate([0, 1, 3, 5, 100], start=2):
    print(f"Question {i}. {v_E(m * tau):.7f}")

for i, m in enumerate([0, 1, 3, 5, 100], start=7):
    print(f"Question {i}. {i_E(m * tau):.7f}")

for i, m in enumerate([0, 1, 3, 5, 100], start=12):
    print(f"Question {i}. {i_s(m * tau):.7f}")
```

### Scenario 4

```python
tau = 0.5e-3
```

```python
def i_E(t):
    return 0.5 + (1.5 * mt.exp(-t / 0.5e-3))

def v_E(t):
    return - 3 * mt.exp(-t / 0.5e-3)

def i_s(t):
    return 1 + i_E(t)
```

```python
for i, m in enumerate([0, 1, 3, 5, 100], start=2):
    print(f"Question {i}. {i_E(m * tau):.7f}")

for i, m in enumerate([0, 1, 3, 5, 100], start=7):
    print(f"Question {i}. {v_E(m * tau):.7f}")

for i, m in enumerate([0, 1, 3, 5, 100], start=12):
    print(f"Question {i}. {i_s(m * tau):.7f}")
```
