## Equation 1

The general solution to
$$
2 \left(3 - 2x\right)^{4} y^{\left(4\right)} + 4 \left(3 - 2x\right)^{3} y^{\prime\prime\prime} + 32 \left(3 - 2x\right)^{2} y^{\prime\prime} + 128 \left(3 - 2x\right) y^{\prime} + 256 y = 0
$$

is
$$
y\left(x\right)
=
C_{1} \left(3 - 2x\right)
+ C_{2} \left(3 - 2x\right)^{2}
+ C_{3} \left(3 - 2x\right)^{2} \ln\left(3 - 2x\right)
+ C_{4} \left(3 - 2x\right)^{2} \ln^{2}\left(3 - 2x\right)
$$

For the initial conditions
- $y \left(1\right) = 1$
- $y^{\prime} \left(1\right) = -2$
- $y^{\prime\prime} \left(1\right) = 8$
- $y^{\prime\prime\prime} \left(1\right) = -48$

the particular solution is
$$
y \left(x\right)
=
\left(3 - 2x\right)
+ \left(3 - 2x\right)^{2} \ln^{2} \left(3 - 2x\right)
$$

For the boundary conditions
- $y \left(1\right) = 1$
- $y^{\prime} \left(1\right) = -2$
- $y \left(1.5 - 0.5e\right) = e + e^{2}$
- $y^{\prime\prime} \left(1.5 - 0.5e\right) = -2 - 8e$

the particular solution is
$$
y \left(x\right)
=
\left(3 - 2x\right)
+ \left(3 - 2x\right)^{2} \ln^{2} \left(3 - 2x\right)
$$

## Equation 2

$$
162 \left(1 - \frac{2}{3} t\right)^{4} w^{\left(4\right)}
- 648 \left(1 - \frac{2}{3} t\right)^{3} w^{\prime\prime\prime}
+ 864 \left(1 - \frac{2}{3} t\right)^{2} w^{\prime\prime}
- 288 \left(1 - \frac{2}{3} t\right) w^{\prime}
+ 128 w
=
0
$$

is
$$
w\left(t\right)
=
C_{1} \cos \left(\ln\left(3 - 2t\right)\right)
+ C_{2} \sin \left(\ln\left(3 - 2t\right)\right)
+ C_{3} \cos \left(2\ln\left(3 - 2t\right)\right)
+ C_{3} \sin \left(2\ln\left(3 - 2t\right)\right)
$$

For the initial conditions
- $w \left(1\right) = 0$
- $w^{\prime} \left(1\right) = -4$
- $w^{\prime\prime} \left(1\right) = 4$
- $w^{\prime\prime\prime} \left(1\right) = 92$

the particular solution is
$$
w \left(t\right)
=
\cos \left(\ln \left(3 - 2t\right)\right)
+ \frac{1}{2} \sin \left(\ln \left(3 - 2t\right)\right)
- \cos \left(2\ln \left(3 - 2t\right)\right)
+ \frac{3}{4} \sin \left(2\ln \left(3 - 2t\right)\right) 
$$

For the boundary conditions
- $w \left(1\right) = 0$
- $w^{\prime} \left(1\right) = -4$
- $w \left(1.5 - 0.5e^{\pi}\right) = -2$
- $w^{\prime} \left(1.5 - 0.5e^{\pi}\right) = -2 e^{\pi}$

the particular solution is
$$
w \left(t\right)
=
\cos \left(\ln \left(3 - 2t\right)\right)
+ \frac{1}{2} \sin \left(\ln \left(3 - 2t\right)\right)
- \cos \left(2\ln \left(3 - 2t\right)\right)
+ \frac{3}{4} \sin \left(2\ln \left(3 - 2t\right)\right)
$$

## Equation 3

The general solution to

$$
\frac{d^{3}F}{dr^{3}}
- \frac{3}{r} \frac{d^{2}F}{dr^{2}}
+ \frac{6}{r^{2}} \frac{dF}{dr}
=
\frac{3}{r^{3}} + \frac{3 \ln r}{r^{3}}
$$

is
$$
F \left(r\right)
=
C_{1}
+ C_{2} r^{3} \cos(\sqrt{2}\ln r)
+ C_{3} r^{3} \sin(\sqrt{2}\ln r)
+ \frac{3}{22} \ln^{2} r
+ \frac{51}{121} \ln r
$$

For the initial conditions
- $F \left(1\right) = 2$
- $F^{\prime} \left(1\right) = \frac{414}{121}$
- $F^{\prime\prime} \left(1\right) = \frac{466}{121}$

the particular solution is
$$
F \left(r\right)
=
1
+ r^{3} \cos(\sqrt{2}\ln r)
+ \frac{3}{22} \ln^{2} r
+ \frac{51}{121} \ln r
$$

For the boundary conditions
- $F \left(1\right) = 2$
- $F^{\prime} \left(1\right) = \frac{414}{121}$
- $F \left(2\right) = 1 + 8 \cos \left(\sqrt{2} \ln 2\right) + \frac{3}{22} \ln^{2} 2 + \frac{51}{121} \ln 2$

the particular solution is
$$
F \left(r\right)
=
1
+ r^{3} \cos(\sqrt{2}\ln r)
+ \frac{3}{22} \ln^{2} r
+ \frac{51}{121} \ln r
$$

## Scenario 1

```python
def y(x):
    u = 3 - (2*x)
    return npp.Polynomial([0, 1, mt.log(u)**2])(u)

def w(t):
    l = mt.log(3 - (2*t))
    return mt.cos(l) + (0.5 * mt.sin(l)) - mt.cos(2 * l) + (0.75 * mt.sin(2 * l))

def F(r):
    l = mt.log(r)
    return 1 + ((r**3) * mt.cos(mt.sqrt(2) * l)) + ((l**2) * 3 / 22) + (l * 51 / 121)
```

```python
indys = np.linspace(-6.9, -4.21, num=5, endpoint=True)

for i, indy in enumerate(indys):
    print(f"Item {i+1}. y({indy}) = {y(indy)}")
for i, indy in enumerate(indys):
    print(f"Item {i+6}. w({indy}) = {w(indy)}")
for i, indy in enumerate(indys):
    print(f"Item {i+11}. F({-indy}) = {F(-indy)}")
```
